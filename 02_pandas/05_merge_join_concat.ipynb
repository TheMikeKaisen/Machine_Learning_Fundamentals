```python
import pandas as pd
import numpy as np
import matplotlib as mlt
```

```python
courses = pd.read_csv('contents/courses.csv')
students = pd.read_csv('contents/students.csv')
nov = pd.read_csv('contents/reg-month1.csv')
dec = pd.read_csv('contents/reg-month2.csv')

matches = pd.read_csv('contents/matches.csv')
delivery = pd.read_csv('contents/deliveries.csv')
```

```python
# 1. CONCAT
```

```python
# nov and dec dataframes are similar and represents similar data for two different months.
regs = pd.concat([nov,dec])

# Using concat, we can stack two dataframes together vertically
regs

# PROBLEM: You might see that after a certain point, indexes repeat. To avoid that there is a parameter called "ignore_index".
regs = pd.concat([nov, dec], ignore_index=True)
regs
```

### MultiIndex DataFrame

```python
multi = pd.concat([nov,dec],keys=['Nov','Dec'])
# Multiindex DataFrame
multi
```

```python
# to print the data for November only
multi.loc['Nov']
```

```python
# To print any index withing Nov index
# pass a tuple of index and sub-index
multi.loc[[('Nov', 4)]] # dataframe result
# multi.loc[('Nov', 4)] # series
```

```python
# to concat two dataframes HORIZONTALLY
pd.concat([nov,dec],axis=1)
```

### JOINS

```python
# inner join
# df1.merge(df2, how="inner/left/right/outer", on="common column")
# inner join consists of common data from both the dataframes
students.merge(regs,how='inner',on='student_id')
```

```python
# left join
# all the data of left dataframe will be present along with the common data from right dataframe
# if right df does not contain certain data, Nan is placed instead
courses.merge(regs,how='left',on='course_id')
```

```python
# right join
# adding some data below students df to understand right join better
temp_df = pd.DataFrame({
    'student_id':[26,27,28],
    'name':['Nitish','Ankit','Rahul'],
    'partner':[28,26,17]
})

students = pd.concat([students,temp_df],ignore_index=True)
```

```python
students.tail()
```

```python
students.merge(regs,how='right',on='student_id')
```

```python
regs.merge(students,how='left',on='student_id')
```

```python
# outer join
students.merge(regs,how='outer',on='student_id').tail(10)
```

```python
# 1. find total revenue generated
comb = regs.merge(courses, how="inner", on="course_id")
total_revenue = comb['price'].sum()
print(total_revenue)
```

```python
# 2. find month by month revenue
temp_df = pd.concat([nov,dec],keys=['Nov','Dec']).reset_index()
temp_df.merge(courses,on='course_id').groupby('level_0')['price'].sum()
```

```python
# 3. Print the registration table
# cols -> name -> course -> price
```

```python
students.merge(regs, how='inner', on='student_id').merge(courses, how='inner', on='course_id')[['name', 'course_name', 'price']]
```

```python
regs.merge(courses, on='course_id')[['course_name', 'price']].groupby('course_name').sum().plot(kind='bar')
```

```python
# 5. find students who enrolled in both the months
common_student_id = np.intersect1d(nov['student_id'],dec['student_id'])
common_student_id
```

```python
students[students['student_id'].isin(common_student_id)]
```

```python
# 6. find course that got no enrollment
# courses['course_id']
# regs['course_id']

course_id_list = np.setdiff1d(courses['course_id'],regs['course_id'])
courses[courses['course_id'].isin(course_id_list)]
```

```python
# 7. find students who did not enroll into any courses
student_id_list = np.setdiff1d(students['student_id'],regs['student_id'])
students[students['student_id'].isin(student_id_list)].shape[0]
```

```python
# 8. Print student name -> partner name for all enrolled students
# self join
students.merge(students,how='inner',left_on='partner',right_on='student_id')[['name_x','name_y']]
```

```python
# 9. find top 3 students who did most number enrollments
regs.merge(students,on='student_id').groupby(['student_id','name'])['name'].count()
```

```python
# 10. find top 3 students who spent most amount of money on courses
regs.merge(students,on='student_id').merge(courses,on='course_id').groupby(['student_id','name'])['price'].sum().sort_values(ascending=False).head(3)
```
